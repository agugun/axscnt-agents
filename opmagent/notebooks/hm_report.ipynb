# History match report

Made by `opmagent hm report` (or the history matcher's `hm_report`). **Read only:** it plots the observed and simulated monthly series and the misfit the HM functions computed; it changes nothing.

```python
payload_path = "payload.json"   # written by opmagent.hm.service.write_report
```

```python
import plotly.io as pio
pio.renderers.default = 'notebook'   # self-contained HTML export
from IPython.display import display
from opmagent.notebooks import hm_report_lib as hm

data = hm.load(payload_path)
hm.header(data)
```

## Findings

The worst terms in words, and water breakthrough simulated against observed.

```python
hm.findings(data)
```

## Parameters and sensitivity

For a history match: the parameters with their ranges and the best case's values (flagged at a bound), the sensitivity tornado (each parameter at its low and high end, the others at their start), and the cases evaluated, best first.

```python
if data.get('parameters'):
    display(hm.parameters_frame(data))
f = hm.fig_tornado(data)
if f is not None:
    display(f)
    display(hm.tornado_frame(data).round(4))
if data.get('cases'):
    display(hm.cases_frame(data).round(4))
else:
    print('A single run: no parameters or cases.')
f = hm.fig_convergence(data)
if f is not None:
    display(f)
```

## Misfit per well and quantity

`misfit` is the mean of ((sim − obs) / σ)² over the months both have. Quantities the history control honours (given to Flow) are shown but left out of the total.

```python
display(hm.wells_frame(data).round(3))
display(hm.terms_frame(data).round(3))
f = hm.fig_misfit_bars(data)
if f is not None:
    display(f)
f = hm.fig_misfit_map(data)
if f is not None:
    display(f)
```

## Field

```python
f = hm.fig_field(data)
if f is not None:
    display(f)
```

## Wells

Observed (circles) against simulated (line), with the measurement band (± σ). Water breakthrough (water cut above 5 %) is marked, observed dotted and simulated dashed.

```python
for w in hm.wells(data):
    f = hm.fig_well(data, w)
    if f is not None:
        display(f)
```
